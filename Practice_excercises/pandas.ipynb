```python
import pandas as pandu
data = {
    'a' :[420,300,350,None],
    'b': [50,None,40,45],
    'c' : [1,2,None,None]

}

pandi = pandu.DataFrame(data)
print("ori data: \n",pandi)

pandi.fillna(pandi.mean(), inplace=False)
```
