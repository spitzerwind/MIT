```python
print("enter two number on which operations would be performed")
num1= int(input("first number"))
          
num2= int(input("second number"))

operationInpt = int(input("choose operation from 1. add 2. subtract 3. multiply 4. divide (provide int value like 1 for addition"))

if operationInpt == 1:
    sum = num1+num2
    print(sum)
elif operationInpt == 2:
    sub = num1-num2
    print(sub)
elif operationInpt == 3:
    mul = num1*num2
    print(mul)
elif operationInpt == 4:
    if num2 != 0:
        div = num1/num2
        print(div)
    else:
        print("not operational invalid input number 0")
else :
    print("wrong operation is chosen")



```
