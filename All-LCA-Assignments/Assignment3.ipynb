```python
# to check if givem=n triangle is right angle triangle while using functions
def checker(a,b,c):
    sum_squarea = a*a + b*b
    sum_squareb = c**2 + b*b
    sum_squarec = a*a + c**2
    if sum_squarea == c**2:
        print("right angle triangle!")
    elif sum_squareb == a**2:
        print("right angle triangle!")
    elif sum_squarec == b**2:
        print("right angle triangle!")
    else:
        print("not an right angle triangle")



num1 = int(input("a:"))
num2 = int(input("b:"))
num3 = int(input("c:"))
checker(num1,num2,num3)
```
