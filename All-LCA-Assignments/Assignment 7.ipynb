```python
# import matplotlib.plt as plt
import pandas as pd

data = pd.read_csv(r"C:\Users\Admin\Downloads\Titanic-Dataset (2).csv")
print(data)

import matplotlib.pyplot as plt
counts = data["Pclass"].value_counts().sort_index()

plt.bar(counts.index ,counts.values)
```

```python
counts = data.groupby(["Survived","Sex"]).size().unstack()
print(counts)

counts.plot(kind="bar")

plt.xlabel("survival status")
plt.ylabel("number of passengers")
plt.title("titanic survival by gen")

plt.xticks([0,1], ["not survives","survived"], rotation=0)
plt.show
```

```python
datas = data.dropna(subset =["Age","Fare"])

plt.scatter(data["Age"],data["Fare"])

plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare in Titanic")

plt.show
```

```python




datas = data.dropna(subset =["Pclass","Fare"])

plt.scatter(data["Pclass"],data["Fare"])

plt.xlabel("passenger class")

plt.ylabel("fare")
plt.title("class vs fare")
plt.xticks([1,2,3])
plt.grid(True)
plt.show




```

```python
fare = data.groupby("Pclass"),["Fare"].mean()
plt.plot(fare.index,fare.values , marker ="o")

plt.xlabel("passenger class")

plt.ylabel("avg fare")
plt.title("avg fare by class ")
plt.xticks([1,2,3])
plt.grid(True)
plt.show
```
